```python
import numpy as np
a = np.memmap('inews.-U', 'H', 'r')
s = np.lib.stride_tricks.as_strided(a[1:1].view('V8'), [len(a)-4], [2]).argsort()
```

```python
def f(x, n=8):
    i = np.searchsorted(
        np.lib.stride_tricks.as_strided(a[1:1].view(f'V{len(x)*2}'), [len(a)-4], [2]),
        np.array([x]).view('H')[-2::-2].ravel().view(f'V{len(x)*2}')[0],
        sorter = s
    )
    i = s[i] - 1
    return a[i:i-n:-1].astype('I').view(f'U{n}')[0]
```

```python
x = '从前有座山, 山里有座庙, 庙里有个老和尚'
{x[:i]: f(x[:i],10) for i in range(1,len(x)+1)}
```
