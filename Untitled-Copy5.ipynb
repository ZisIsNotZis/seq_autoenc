```python
%matplotlib notebook
import torch
import numpy as np
import matplotlib.pyplot as plt
from tqdm.notebook import tqdm
from torch.nn import *

class EmbeddingAuto(EmbeddingBag):
    def __init__(s, n, k, l):
        super().__init__(n, k, mode='sum')
        s.l = Parameter(.5 ** torch.arange(l-1, -1., -1))
    def forward(s, x):
        return super().forward(x, per_sample_weights=s.l.expand_as(x))

class Log:
    n = 0
    def __init__(s, k=1):
        s.a = np.empty([2100,k], 'f')
    def __call__(s, *x):
        s.a[s.n] = x
        s.n += 1
        if not s.n % 10:
            plt.cla()
            plt.plot(s.a[:s.n])
            plt.gcf().canvas.draw()

torch.set_default_tensor_type(torch.cuda.FloatTensor)
a = np.memmap('inews.B')
a = torch.tensor(a).as_strided([len(a)-4,5], [1,1])
```

```python
m = Sequential(
    EmbeddingAuto(256, 8192, 4),
    ReLU(1),
    Linear(8192, 256),
)
o = torch.optim.Adam(m.parameters())
    O = torch.optim.lr_scheduler.OneCycleLR(o, .001, len(a)*20//2048, pct_start=.01, div_factor=99, final_div_factor=99)
log = Log()
for t in tqdm(range(O.total_steps//100)):
    for t in range(100):
        t = a[torch.randint(len(a), [2048])]
        x = m(t[:,:-1].long())
        y = t[:,-1].long()
        functional.cross_entropy(x, y).backward()
        o.step()
        o.zero_grad()
        O.step()
    log((x.argmax(1)==y).sum().cpu()/len(y))
```
