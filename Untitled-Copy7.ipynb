```python
%matplotlib notebook
import torch
import numpy as np
from torch.nn import *
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt

class EmbeddingAuto(EmbeddingBag):
    def __init__(s, n, k, l):
        super().__init__(n, k, mode='sum')
        s.l = Parameter(.5 ** torch.arange(l-1, -1., -1))
    def forward(s, x):
        return super().forward(x, per_sample_weights=s.l.expand_as(x))

class F(Module):
    def __init__(s, f, a=None):
        super().__init__()
        s.f = f
        s.a = a
    def forward(s, x):
        return s.f(x) if s.a is None else s.f(x, s.a)

class Enc(Sequential):
    def forward(s, x):
        while x.size(2) > 1:
            x = super().forward(x)
        return x[...,0]

class Add(Sequential):
    def forward(s, x):
        return s[0](x) + s[1](x)

class Log:
    n = 0
    def __init__(s, n=2100, k=1, m=10):
        s.a = np.empty([2100,k], 'f')
        s.m = m
    def __call__(s, *x):
        s.a[s.n] = x
        s.n += 1
        if not s.n % s.m:
            plt.cla()
            plt.plot(s.a[:s.n])
            plt.gcf().canvas.draw()
torch.set_default_tensor_type(torch.cuda.FloatTensor)
a = np.memmap('../nlpdat/py/all.90', mode='r')
a = torch.tensor(a).as_strided([len(a)-32,33], [1,1])
```

```python
k = 64
K = 64
m = Sequential(
    Embedding(90, k),
    F(lambda x:x.transpose(1,2)),
    Enc(
        Add(
            Sequential(
                Conv1d(k, k, 3, padding=1, bias=0),
                BatchNorm1d(k, track_running_stats=0),
                ReLU(1),
            ),
            Conv1d(k, k, 3, padding=1, bias=0),
        ),
        MaxPool1d(2),
    ),
    Linear(k, K),
    ReLU(1),
    Linear(K, 90),
)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, .01, len(a)//8192, pct_start=.1, div_factor=9, final_div_factor=999)
log = Log()
for x in tqdm(range(O.total_steps//50)):
    for x in range(50):
        x = a[torch.randint(len(a), [8192])]
        y = x[:,-1].long()
        x = m(x[:,:-1].long())
        functional.cross_entropy(x, y).backward() 
        o.step()
        o.zero_grad()
        O.step()
    log((x.argmax(1)==y).sum().cpu()/len(y))
```

```python
k = 512
K = 512
m = Sequential(
    Embedding(90, k),
    F(lambda x:x.transpose(1,2)),
    Enc(
        Add(
            Sequential(
                Conv1d(k, k, 3, padding=1, bias=0),
                BatchNorm1d(k, track_running_stats=0),
                ReLU(1),
            ),
            Conv1d(k, k, 3, padding=1, bias=0),
        ),
        Add(
            Sequential(
                Conv1d(k, k, 3, padding=1, bias=0),
                BatchNorm1d(k, track_running_stats=0),
                ReLU(1),
                Conv1d(k, k, 1, bias=0),
            ),
            Conv1d(k, k, 3, padding=1, bias=0),
        ),
        MaxPool1d(2),
    ),
    Linear(k, K),
    ReLU(1),
    Linear(K, 90),
)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, .0005, len(a)//1024, pct_start=.01, div_factor=9, final_div_factor=99)
log = Log(m=1)
for i in tqdm(range(1, O.total_steps)):
    x = a[torch.randint(len(a), [1024])]
    y = x[:,-1].long()
    x = m(x[:,:-1].long())
    functional.cross_entropy(x, y).backward() 
    o.step()
    o.zero_grad()
    O.step()
    if not i % 500:
        log((x.argmax(1)==y).sum().cpu()/len(y))
```
