```python
import numpy as np
m = np.memmap("inews.7622", "H", 'r')
```

```python
from collections import defaultdict, Counter
n = 2
m = np.lib.stride_tricks.as_strided(m, (len(m)-n,n+1), (2,2))
a = defaultdict(Counter)
for i,j in zip(np.sort(m[:,:-1]).view(f'S{n*2}')[:,0], m[:,-1]):
    a[i][j] += 1
a = np.fromiter(((i,j.most_common(1)[0][0])for i,j in a.items()), f'V{n*2},H')
a.sort()
a.tofile(f'inews.-{n}H.H')
(a['f1'][np.searchsorted(a['f0'],np.sort(m[:,:-1]).view(f'V{n*2}')[:,0])]==m[:,-1]).sum()/len(m)
```
